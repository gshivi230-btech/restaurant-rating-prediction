```python
df_clean = df.copy()
```

```python
df_clean = df_clean.drop(columns=["Rating color", "Rating text"])
```

```python
df_clean = df_clean.drop(columns=["Restaurant ID"])
```

```python
df_clean = df_clean.drop(
    columns=["Restaurant Name", "Address", "Locality Verbose"]
)
```

```python
df_clean.columns.tolist()
```

```python
df_clean["Cuisines"] = df_clean["Cuisines"].fillna("Unknown")
```

```python
df_clean.isnull().sum()
```

```python
df_clean.shape
```

```python
y = df_clean["Aggregate rating"]
```

```python
X = df_clean.drop(columns=["Aggregate rating"])
```

```python
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
categorical_features = X.select_dtypes(include=["object"]).columns.tolist()

categorical_features
```

```python
numerical_features = X.select_dtypes(include=["int64", "float64"]).columns.tolist()

numerical_features
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

preprocessor = ColumnTransformer(
    transformers=[
        ("num", "passthrough", numerical_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ]
)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

linear_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", LinearRegression())
    ]
)

linear_model.fit(X_train, y_train)
```

```python
y_pred_linear = linear_model.predict(X_test)
```

```python
y_pred_linear[:10]
```

```python
comparison = pd.DataFrame({
    "Actual Rating": y_test.values[:10],
    "Predicted Rating": y_pred_linear[:10]
})

comparison
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np
```

```python
mae_linear = mean_absolute_error(y_test, y_pred_linear)
mae_linear
```

```python
mse_linear = mean_squared_error(y_test, y_pred_linear)
mse_linear
```

```python
rmse_linear = np.sqrt(mse_linear)
rmse_linear
```

```python
r2_linear = r2_score(y_test, y_pred_linear)
r2_linear
```

```python
print("Linear Regression Results")
print("-------------------------")
print("MAE  :", mae_linear)
print("MSE  :", mse_linear)
print("RMSE :", rmse_linear)
print("R²   :", r2_linear)
```

```python
from sklearn.tree import DecisionTreeRegressor

decision_tree_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", DecisionTreeRegressor(
            max_depth=10,
            random_state=42
        ))
    ]
)
```

```python
decision_tree_model.fit(X_train, y_train)
```

```python
y_pred_tree = decision_tree_model.predict(X_test)
```

```python
mae_tree = mean_absolute_error(y_test, y_pred_tree)
mse_tree = mean_squared_error(y_test, y_pred_tree)
rmse_tree = np.sqrt(mse_tree)
r2_tree = r2_score(y_test, y_pred_tree)

print("Decision Tree Results")
print("--------------------")
print("MAE  :", mae_tree)
print("MSE  :", mse_tree)
print("RMSE :", rmse_tree)
print("R²   :", r2_tree)
```

```python
from sklearn.ensemble import RandomForestRegressor

random_forest_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", RandomForestRegressor(
            n_estimators=100,
            max_depth=10,
            random_state=42,
            n_jobs=-1
        ))
    ]
)
```

```python
random_forest_model.fit(X_train, y_train)

y_pred_forest = random_forest_model.predict(X_test)

mae_forest = mean_absolute_error(y_test, y_pred_forest)
mse_forest = mean_squared_error(y_test, y_pred_forest)
rmse_forest = np.sqrt(mse_forest)
r2_forest = r2_score(y_test, y_pred_forest)

print("Random Forest Results")
print("--------------------")
print("MAE  :", mae_forest)
print("MSE  :", mse_forest)
print("RMSE :", rmse_forest)
print("R²   :", r2_forest)
```

```python
results = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Decision Tree",
        "Random Forest"
    ],
    "MAE": [
        mae_linear,
        mae_tree,
        mae_forest
    ],
    "MSE": [
        mse_linear,
        mse_tree,
        mse_forest
    ],
    "RMSE": [
        rmse_linear,
        rmse_tree,
        rmse_forest
    ],
    "R²": [
        r2_linear,
        r2_tree,
        r2_forest
    ]
})

results
```

# Conclusion

This project developed machine learning models to predict restaurant aggregate ratings using restaurant-related features.

Three regression models were implemented:

1. Linear Regression
2. Decision Tree Regressor
3. Random Forest Regressor

The models were evaluated using:

- MAE
- MSE
- RMSE
- R² Score

Random Forest achieved the strongest test-set performance among the three models:

- MAE = 0.2006
- MSE = 0.0946
- RMSE = 0.3075
- R² = 0.9585

This indicates that the model explains approximately 95.85% of the variation in restaurant ratings on the test set.

The analysis suggests that restaurant attributes such as votes, price range, cuisines, city, and service-related features contribute significantly to rating prediction.
